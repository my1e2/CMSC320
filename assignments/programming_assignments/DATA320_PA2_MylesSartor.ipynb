# 1.2. Functions

Run the cell below.

```python
max(l_10)
```

`max(x)` is a built-in function in Python, but we can build our own. Here is a custom function that returns the max of a list. Run the cell below so the fuction is stored in memory.

```python
def custom_max(x):
  m = x[0] #m is first defined to be the first element of list x
  for i in x: #for each element in list x
    if i > m: #if that element is greater than m
      m = i #make m to be now equal to that element

  return m #after the iterations are done, return m
```

Now, run the cell below to use the function on `l_10`.

```python
custom_max(l_10)
```

12. Is it possible/easy to do Blinding for this experiment?

ANSWER Question 12 HERE

Blinding in this experiment might not be easy to execute, but it would certainly be possible if students that were studying for good test scores were not aware of what study plan they were under beforehand.
