## 2. Data cleaning

**2.1 (10 points):** We want to be able to use the rows with no provided eye color. What we'll do is fill those missing values with something.

***Please read the directions carefully:***


- In order to fill in a missing eye color for someone, we want to find the **most common eye color** among people who have the **same hair and skin color** as the person.

- For instance, if we want to fill in the eye color of someone who has yellow hair and violet skin, we replace the missing eye color with the eye color that we identified to be the most common in people with yellow hair and violet skin.

**HINT:** you may find it useful to first find the most common eye color for each combination of hair and skin colors so you don't do that more than once for each combination of hair and skin features.

**NOTE:** in case you find multiple modes for some group, choose the color that appears first in alphabetical order.

When you finish filling in the missing eye colors, please display **profiles_df**.


```python
# Write your code here
# Remember to feel free to add code cells as needed

profiles_df['Eye_Color'] = profiles_df.groupby(
['Hair_Color','Skin_Color'])['Eye_Color'].transform(lambda x: x.fillna(x.mode()[0] if not x.mode().empty else x))

# defining and using (transforming) lambda function to fill in most common eye colors for each combination of hair and skin colors

profiles_df
```

## 5. Finding who might own the ticket

**5.1 (1 point):** Load **"Zichao_s Magic Show.csv"** into a dataframe called **"show_df"** and display it when you are done.

```python
# Write your code here

show_df = pd.read_csv('Zichao_s Magic Show.csv')
show_df
```


**5.2 (8 points):**  Remember that there was a show ticket left at the crime scene, and it'd be helpful to know which tickets match the color of the ticket at the scene.

Create a column called **Ticket_Color** for the **"show_df"** that displays the colors of each ticket based on its Ticket ID.

How do you find the ticket color from the ticket number to populate the new column? **You asked the ticket issuer and learned the rule**: Sum the individual digits in a ticket number, then look at the rightmost digit and use this mapping:

0-2: Blue, 3-5: Red, 6-7: Green, 8-9: Yellow.

Display the **"show_df"** when you are done.

```python
# Write your code here

def ticket_color(ticket):
    digit_sum = sum(int(d) for d in str(ticket)) # looping through each digit as a string before converting to integer and summing them up
    rightmost = digit_sum % 10 # right most digit following summing of the ticket number (modulus remainder will always be the right digit)
    
    if rightmost <= 2: # 0-2
        return "Blue"
    elif rightmost <= 5: # 3-5
        return "Red"
    elif rightmost <= 7: # 6-7
        return "Green"
    else:
        return "Yellow" # 8-9

show_df["Ticket_Color"] = show_df["Ticket_Number"].apply(ticket_color) # applying function defined to each ticket number and storing result in new ticket color column
show_df

# function to specify ticket color rule

# confirming that the ticket number is converted to integer

# summing integers/each digit up
```

**5.2 (3 points):** Add the ticket color to the **Profiles** dataframe. It should retain all the **Profiles** rows, and include a ticket color for the row if that person has a ticket. Display **Profiles**.

```python
profiles_df = profiles_df.merge(show_df[['ID','Ticket_Color']], on='ID', how='left')

profiles_df # left join to get every row from profiles table and only matches from ticket show table (if the person has a ticket the color is display, if not the color is 'NaN')
```
