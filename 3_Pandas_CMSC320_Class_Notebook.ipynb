# How to Filter a Column based on Condition(s)

To filter a column in a pandas DataFrame, we can use boolean indexing. <br>



> **Syntax: DataFrame['column name'][boolean condition]**

E.g.: Filter only the 'Name' column for values equal to 'Jonas Jerebko'


```python
df['Name'][df['Name']=='Jonas Jerebko']
```

```python
print(df['Name'][df['Name']=='Jonas Jerebko'])
```

# How to Filter Rows based on Condition(s)



> **Syntax: DataFrame[boolean condition]**


E.g.: Filter rows where age is greater than 25.

```python
df[df['Age']>25]
```

```python
print(df[df['Age']>25])
```

**Q: What about multiple conditions?**

* E.g.: Filter rows where age is greater than 25 and Weight > 210

```python
df[(df['Age']>25) & (df['Weight']>210)]
```

```python
# how many times each unique values appears in a series/column
df['Age'].value_counts()
```
