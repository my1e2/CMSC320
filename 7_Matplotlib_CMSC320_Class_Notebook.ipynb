
# Data Visualization with Matplotlib

This notebook introduces data visualization using Matplotlib, focusing on understanding and interpreting different types of plots.  
We’ll use the **Diabetes dataset** from `sklearn.datasets` and assume you already have a basic understanding of **Pandas**.

- Create and interpret basic visualizations with Matplotlib

- Use subplots to display multiple visualizations
- Visualize relationships between variables in the Diabetes dataset


```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn import datasets

%matplotlib inline
```

## 1. Loading the Diabetes Dataset

```python

# Load the diabetes dataset
diabetes = datasets.load_diabetes()

# Convert to a Pandas DataFrame
df = pd.DataFrame(diabetes.data, columns=diabetes.feature_names)
df['target'] = diabetes.target

# Display first few rows
df.head()
```


## 2. Line Plot

A **line plot** is used to display trends over an index or continuous variable.  
It’s useful when visualizing **how a value changes** — for instance, blood sugar levels across patients.

Below, we plot the first 50 patient indices against their diabetes progression target value.


```python

plt.figure(figsize=(8,5))
plt.plot(df.index[:50], df['target'][:50], marker='o', color='teal')
plt.title('Diabetes Progression (First 50 Patients)')
plt.xlabel('Patient Index')
plt.ylabel('Disease Progression Measure')
plt.show()
```


## 3. Histogram

A **histogram** shows the distribution of a continuous variable.  
It divides data into **bins** and shows how many data points fall into each bin.

Below, we visualize the distribution of **Body Mass Index (BMI)** across patients.


```python

plt.figure(figsize=(7,5))
plt.hist(df['bmi'], bins=20, color='skyblue', edgecolor='black')
plt.title('Distribution of BMI (Body Mass Index)')
plt.xlabel('BMI')
plt.ylabel('Frequency')
plt.show()
```

```python
plt.figure(figsize=(7,5))
plt.hist(df['sex'])
```


## 4. Scatter Plot

A **scatter plot** is used to show the relationship between two continuous variables.  
It’s great for detecting **correlations or patterns** between features.

Here we examine how **BMI** relates to the **disease progression target**.


```python

plt.figure(figsize=(7,5))
plt.scatter(df['bmi'], df['target'], alpha=0.7, color='salmon')
plt.title('BMI vs. Diabetes Progression')
plt.xlabel('Body Mass Index (bmi)')
plt.ylabel('Progression (target)')
plt.show()
```


## 5. Boxplot

A **boxplot** summarizes the distribution of data by showing:
- The **median** (center line)
- The **quartiles** (the box edges)
- **Outliers** (points beyond the whiskers)

Below, we create a boxplot for **Blood Pressure (bp)**.


```python

plt.figure(figsize=(6,5))
plt.boxplot(df['bp'], vert=True, patch_artist=True, boxprops=dict(facecolor='lightgreen'))
plt.title('Boxplot of Blood Pressure')
plt.ylabel('Blood Pressure (bp)')
plt.show()
```


## 6. Bar Plot

A **bar plot** displays quantities for discrete categories.  
Since the Diabetes dataset has continuous features, we can group patients by **above-average** and **below-average BMI** and compare their **mean progression**.


```python

df['bmi_group'] = np.where(df['bmi'] > df['bmi'].mean(), 'Above Avg BMI', 'Below Avg BMI')
avg_progression = df.groupby('bmi_group')['target'].mean()

plt.figure(figsize=(6,4))
plt.bar(avg_progression.index, avg_progression.values, color=['orange', 'lightblue'])
plt.title('Average Disease Progression by BMI Group')
plt.ylabel('Average Progression (target)')
plt.show()
```


## 7. Subplots

**Subplots** let you display multiple visualizations side-by-side for comparison.  
Below we show histograms of **BMI** and **blood pressure (bp)** in one figure.


```python

fig, axes = plt.subplots(1, 2, figsize=(10,4))

axes[0].hist(df['bmi'], bins=15, color='lightcoral', edgecolor='black')
axes[0].set_title('Distribution of BMI')

axes[1].hist(df['bp'], bins=15, color='lightseagreen', edgecolor='black')
axes[1].set_title('Distribution of Blood Pressure')

plt.suptitle('Comparing Two Key Health Features')
plt.tight_layout()
plt.show()
```


## 8. Customizing Your Plots

Matplotlib allows extensive customization — you can adjust:
- Colors (`color`, `cmap`)
- Titles (`plt.title()`)
- Axis labels (`plt.xlabel()`, `plt.ylabel()`)
- Legends (`plt.legend()`)
- Styles (`plt.style.use()`)

Let’s customize a scatter plot with more visual elements.


```python

plt.figure(figsize=(7,5))
plt.scatter(df['s5'], df['target'], color='purple', alpha=0.7, label='Patients')
plt.title('Serum Measurement (s5) vs. Diabetes Progression')
plt.xlabel('Serum s5')
plt.ylabel('Progression Target')
plt.legend()
plt.grid(True)
plt.show()
```


## 9. Plotting Directly from Pandas

Pandas integrates smoothly with Matplotlib, allowing you to create plots directly from DataFrames using `.plot()`.


```python

df[['bmi', 'bp']].plot(kind='line', figsize=(8,5), title='Pandas Line Plot Example')
plt.show()
```
