
# Neural Networks for Digit Recognition

We will build a neural network to recognize handwritten digits using the MNIST dataset.



## Reminder: What is a Neural Network?

A neural network is a model inspired by the human brain. It consists of layers of "neurons" that transform input data.

- Input layer: receives data (pixels)
- Hidden layers: learn patterns
- Output layer: predicts digit (0–9)

```
Input (784 pixels (each image is 28*28, greyscale)
     ↓
[ Dense Layer ]
     ↓
[ Dense Layer ]
     ↓
Output (10 classes)
```


## Import Libraries

````python

import numpy as np
import matplotlib.pyplot as plt
from tensorflow import keras
from tensorflow.keras import layers
````

## Load Dataset (MNIST)

````python

(x_train, y_train), (x_test, y_test) = keras.datasets.mnist.load_data()

print("Training shape:", x_train.shape)
print("Test shape:", x_test.shape)
````

## Visualize the Data

````python

plt.imshow(x_train[0], cmap='gray')
plt.title(f"Label: {y_train[0]}")
plt.show()
````


## Preprocess the Data

We:
- Normalize pixel values (0–255 → 0–1)
- Flatten images (28×28 → 784)


````python

x_train = x_train / 255.0
x_test = x_test / 255.0

x_train = x_train.reshape(-1, 784)
x_test = x_test.reshape(-1, 784)
````

````python
x_train
````


## Interactive: Change the Network

We have 2 layers. Wecan change:
- The number of neurons in each layer
- The number of epochs


````python

# Try changing these values!
layer1_units = 128
layer2_units = 64
epochs = 5
````

## Build the Model

````python
#we can also change the activation functions here:
model = keras.Sequential([
    layers.Dense(layer1_units, activation='relu', input_shape=(784,)),
    layers.Dense(layer2_units, activation='relu'),
    layers.Dense(10, activation='softmax')
])

model.summary()
````

## Compile the Model

````python

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)
````

## Train the Model

````python

history = model.fit(
    x_train, y_train,
    epochs=epochs,
    batch_size=32,
    validation_split=0.1
)
````

## Evaluate the Model

````python

test_loss, test_acc = model.evaluate(x_test, y_test)
print("Test accuracy:", test_acc)
````

## Make Predictions

````python
predictions = model.predict(x_test)

i = 0
plt.imshow(x_test[i].reshape(28,28), cmap='gray')
plt.title(f"Predicted: {np.argmax(predictions[i])}, Actual: {y_test[i]}")
plt.show()
````

## Training Visualization

````python
print(history.history['accuracy'])
````

````python
epochs = range(1, len(history.history['accuracy']) + 1)
plt.xticks(epochs)

plt.plot(epochs, history.history['accuracy'], label='train')
plt.plot(epochs, history.history['val_accuracy'], label='validation')

plt.title('Model Accuracy')
plt.ylabel('Accuracy')
plt.xlabel('Epoch')
plt.legend()
plt.show()
````
