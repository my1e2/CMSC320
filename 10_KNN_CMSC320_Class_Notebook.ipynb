# Introduction to K-Nearest Neighbors (KNN)
In this demo, we will explore the **K-Nearest Neighbors (KNN)** algorithm using the Titanic dataset provided by `scikit-learn`.

### Learning Objectives
1. Understand how KNN works
2. Learn how to choose the best value of **K**
3. Evaluate the model using metrics

```python
import pandas as pd
```

## 1. Load the Titanic Dataset

```python
from sklearn.datasets import fetch_openml
titanic = fetch_openml(name='titanic', version=1, as_frame=True)
df = titanic.frame
df.head()
```

## 2. Data Preprocessing
We'll handle missing values, encode categorical variables, and prepare features for KNN.

```python
# Drop irrelevant columns
df = df.drop(['cabin', 'boat', 'body', 'home.dest', 'name', 'ticket', 'embarked'], axis=1)

# Handle missing values
### obsolete
###df['age'].fillna(df['age'].median(), inplace=True)
df['age'] = df['age'].fillna(df['age'].median())

### obsolete
### df.dropna(subset=['fare'], inplace=True)
df = df.dropna(subset=['fare'])

df
```

```python
# Encode categorical variables
### be careful is using
### df = pd.get_dummies(df)
df = pd.get_dummies(df, drop_first=True)

df
```

```python
X = df.drop('survived_1', axis=1)
y = df['survived_1']
```

```python
X
```

```python
y
```

## 3. Train-Test Split and Scaling

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score, confusion_matrix, classification_report

scaler = StandardScaler()
X_scaled = scaler.fit_transform(X)
```

```python
X_scaled
```

```python
X_main, X_test, y_main, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=40)
```

```python
X_train, X_val, y_train, y_val = train_test_split(X_main, y_main, test_size=0.2, random_state=40)
```

```python
print(X.shape)
print(X_train.shape)
print(X_val.shape)
print(X_test.shape)

print(X_train.shape[0]+X_val.shape[0]+X_test.shape[0])
```

## 4. Train a Basic KNN Model

```python
knn = KNeighborsClassifier(n_neighbors=5)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_val)

print('Accuracy:', accuracy_score(y_val, y_pred))
```

## 5. Model Evaluation

```python
print('Confusion Matrix:\n', confusion_matrix(y_val, y_pred))
print('\nClassification Report:\n', classification_report(y_val, y_pred))
```

## 6. Choosing the Best K Value

```python
import matplotlib.pyplot as plt

k_values = range(1, 21)
accuracies = []

for k in k_values:
    knn = KNeighborsClassifier(n_neighbors=k)
    knn.fit(X_train, y_train)
    y_pred = knn.predict(X_val)
    accuracies.append(accuracy_score(y_val, y_pred))

plt.plot(k_values, accuracies, marker='o')
plt.title('Accuracy vs K Value')
plt.xlabel('K')
plt.ylabel('Accuracy')
plt.grid(True)
plt.show()
```

## Final Model Evaluation

```python
knn = KNeighborsClassifier(n_neighbors=3)
knn.fit(X_train, y_train)
y_pred = knn.predict(X_test)

print('Accuracy:', accuracy_score(y_test, y_pred))
print('\nFinal Classification Report:\n', classification_report(y_test, y_pred))
```
