# Grouping and Joining in Pandas


## 1. Grouping Logic: Split-Apply-Combine
The GroupBy process partitions the data based on keys, applies a function to each partition, and then concatenates the results into a new object.



```python
import pandas as pd
import numpy as np

# Constructing the primary dataset
df = pd.DataFrame({
    'Employee': ['Alice', 'Bob', 'Charles', 'Dyvia', 'Eva', 'Faraz'],
    'Dept': ['Sales', 'ENG', 'ENG', 'HR', 'Sales', 'HR'],
    'Project': ['A', 'B', 'A', 'B', 'A', 'B'],
    'Salary': [70, 95, 110, 60, 85, 62],
    'Hours_Week': [40, 45, 50, 35, 42, 38]
})

df
```

```python
df_new_rows = pd.DataFrame({
    'Employee': ['Grace', 'Henry'],
    'Dept': ['ENG', 'Sales'],
    'Project': ['C', 'C'],
    'Salary': [105, 72],
    'Hours_Week': [48, 40]
})


df = pd.concat([df, df_new_rows], axis=0, ignore_index=True)

df
```

```python
# .groupby('Dept') creates a GroupBy object
# ['Salary'] selects the column
# .mean() is the function of interest
avg_salary = df.groupby('Dept')['Salary'].mean()
print("Mean Salary per Department:")
avg_salary
```

```python
multi_group = df.groupby(['Dept', 'Project'])['Salary'].mean()
print("\nMean Salary by Dept and Project:")
multi_group
```

## 2. Advanced Multi-Function Aggregation
The `.agg()` method allows for applying different reduction functions to different features simultaneously.

```python
summary = df.groupby('Dept').agg({
    'Salary': ['min', 'max', 'mean'],  # Returns three statistics for Salary
    'Hours_Week': 'sum'               # Returns total sum for Hours_Week
})
print("Multi-functional Aggregation Summary:")
display(summary)
```

## 3. Pivot Tables for Multidimensional Analysis
Pivot tables provide a way to cross-tabulate data for easier visual comparison.

```python
# Pivot the data to view Salary by Department (Rows) and Project (Columns)
# fill_value=0 handles intersections with no data points
pivot = df.pivot_table(values='Salary', index='Dept', columns='Project', aggfunc='mean', fill_value=0)
print("Pivot Table Analysis:")
display(pivot)
```

```python
#Sort of similar to
multi_group = df.groupby(['Dept', 'Project'])['Salary'].mean()
print("\nMean Salary by Dept and Project:")
multi_group
```

## 4. Broadcasting Statistics with `.transform()`
The `.transform()` method applies a function to each group but returns an object that matches the shape of the original DataFrame.

```python
# Calculate the mean Salary for a row's specific Department and broadcast it back to every row
df['Dept_Avg'] = df.groupby('Dept')['Salary'].transform('mean')

# Calculate the deviation of each employee's salary from their department's mean
df['Salary_Delta'] = df['Salary'] - df['Dept_Avg']

print("DataFrame with Broadcasted Group Statistics:")
display(df[['Employee', 'Dept', 'Salary', 'Dept_Avg', 'Salary_Delta']])
```

## 5. JOINS


```python
df_left = pd.DataFrame({
    'Emp_ID': [1, 2, 3, 4],
    'Name': ['Alice', 'Bob', 'Charlie', 'David'],
    'Dept_ID': [10, 20, 30, 99]
})

df_right = pd.DataFrame({
    'Dept_ID': [10, 20, 30, 40],
    'Dept_Name': ['Sales', 'Engineering', 'HR', 'Marketing'] # Marketing has no employees yet
})

print("Left Table (Employees):")
display(df_left)
print("\nRight Table (Departments):")
display(df_right)
```

## 5.1. Inner Join
Returns only the records where the join key exists in **both** DataFrames. Rows with keys that do not have a match in the other table are discarded.



```python
inner_result = pd.merge(df_left, df_right, on='Dept_ID', how='inner')

inner_result
```

## 5.2. Left Join (Left Outer Join)
Returns all records from the left DataFrame and the matched records from the right DataFrame. If there is no match, the result is `NaN` on the right side.



```python
left_result = pd.merge(df_left, df_right, on='Dept_ID', how='left')

left_result
```

## 5.3. Right Join (Right Outer Join)
Returns all records from the right DataFrame and the matched records from the left DataFrame. If there is no match, the result is `NaN` on the left side.



```python
right_result = pd.merge(df_left, df_right, on='Dept_ID', how='right')

right_result
```

## 5.4. Outer Join (Full Outer Join)
Returns all records when there is a match in either the left or right DataFrame. This effectively creates a union of both tables.



```python
outer_result = pd.merge(df_left, df_right, on='Dept_ID', how='outer')

outer_result
```

## 5.5. Handling Non-Matching Key Names
If the key columns have different names in each DataFrame, use `left_on` and `right_on`.

```python
#original tables
print("Left Table (Employees):")
display(df_left)
print("\nRight Table (Departments):")
display(df_right)

#regular left
left_result = pd.merge(df_left, df_right, on='Dept_ID', how='left')

left_result
```

```python
#renaming one column
df_right_alt = df_right.rename(columns={'Dept_ID': 'Department_Identifier'})

print("Left Table (Employees):")
display(df_left)
print("\nNEW Right Table (Departments):")
display(df_right_alt)
```

```python
alt_merge = pd.merge(df_left, df_right_alt, left_on='Dept_ID', right_on='Department_Identifier', how='inner')

alt_merge
```
